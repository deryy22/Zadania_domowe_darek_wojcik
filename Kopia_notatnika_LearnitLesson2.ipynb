```python
# Informacje o pamięci RAM
print("=== PAMIĘĆ RAM ===")
ram_gb = psutil.virtual_memory().total / (1024**3)
ram_available_gb = psutil.virtual_memory().available / (1024**3)
print(f"Całkowita RAM: {ram_gb:.3f} GB")
print(f"Dostępna RAM: {ram_available_gb:.5f} GB")
print()
```

```python
# 1. ZAWSZE zapisuj ważne wyniki do Google Drive
# 2. Regularnie zapisuj notatnik (Ctrl+S)
# 3. Monitoruj zużycie zasobów
# 4. Zamykaj niepotrzebne sesje
# Przykład: funkcja do automatycznego zapisu wyników
import json
from datetime import datetime
def save_results_to_drive(results_dict, experiment_name):
  """
  Zapisuje wyniki eksperymentu do Google Drive z timestampem
  Args:
    results_dict: Słownik z wynikami eksperymentu
    experiment_name: Nazwa eksperymentu
  """
  timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
  filename = f"/content/drive/MyDrive/DataScience_Projects/results_{experiment_name}_{timestamp}.json"
  with open(filename, 'w') as f:
    json.dump(results_dict, f, indent=2)
  print(f"✓ Wyniki zapisane: {filename}")
  return filename
# Przykład użycia
experiment_results = {
  'model': 'LinearRegression',
  'accuracy': 0.87,
  'mse': 0.13,
  'training_time': '5.2s'
}
save_results_to_drive(experiment_results, 'first_model')
# (odkomentuj gdy Drive jest zamontowany)
```
