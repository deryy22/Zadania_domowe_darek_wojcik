```python
# zadanie 8
sprzedaz = {
(2024, 'Q1'): 120_000,
(2024, 'Q2'): 140_000,
(2024, 'Q3'): 135_000,
(2024, 'Q4'): 160_000
}
print(sprzedaz.get((2024,'Q2'),'Brak Danych'))
sprzedaz[(2025, 'Q1')] = 130_000
print(sprzedaz)
najwyzsza_sprzedaz = max(sprzedaz.values())
for klucz, wartosc in sprzedaz.items():
    if wartosc == najwyzsza_sprzedaz:
       okres_max = klucz
       break
print(f"Najwyzsza wartosc sprzedazy to {najwyzsza_sprzedaz} z okresu: {okres_max}")

najnizsza_sprzedaz = min(sprzedaz.values())
for klucz, wartosc in sprzedaz.items():
    if wartosc == najnizsza_sprzedaz:
       okres_min = klucz
       break
print(f"Najwyzsza wartosc sprzedazy to {najnizsza_sprzedaz} z okresu: {okres_min}")

total_sprzedaz = sum(sprzedaz.values())
print(f"Total sprzedaz z wszystkich okresow wynosi: {total_sprzedaz}")

zbior_kwartalow = set()
okresy = list(sprzedaz.keys())
print(okresy)
for i in range(len(okresy)):
  zbior_kwartalow.add(okresy[i][1])
print(zbior_kwartalow)

#inny sposob
okresy_q = [okres[1] for okres in okresy]
zbior_q = set(okresy_q)
print(zbior_q)
print(len(zbior_q))
```

```python
raport = {
'Q1': {'przychod': 120_000, 'koszt': 80_000},
'Q2': {'przychod': 145_000, 'koszt': 92_000},
'Q3': {'przychod': 138_000, 'koszt': 88_000},
'Q4': {'przychod': 167_000, 'koszt': 101_000}
}

print(f"Przychod za Q1 = {raport.get('Q1').get('przychod')}")
print(f"Przychod za Q2 = {raport.get('Q2').get('przychod')}")
print(f"Przychod za Q3 = {raport.get('Q3').get('przychod')}")
print(f"Przychod za Q4 = {raport.get('Q4').get('przychod')}")

lista_kwartalow = list(raport.keys())

lista_zyskow = []
for i in lista_kwartalow:
  zysk = raport.get(i).get('przychod') - raport.get(i).get('koszt')
  raport[i]['zysk'] = zysk
  lista_zyskow.append(zysk)
  print(f"{i} | {raport[i]['przychod']} | {raport[i]['koszt']:<6} | {raport[i]['zysk']}")

print(f"Suma zyskow: {sum(lista_zyskow)}")
print(f"Minimalny zysk: {min(lista_zyskow)}")
print(f"Maxymalny zysk: {max(lista_zyskow)}")
print(f"Sredni zysk: {sum(lista_zyskow)/len(lista_zyskow)}")
```

```python
# zadanie 17
wersja_1 = {'lr': 0.01, 'epochs': 100, 'optimizer': 'sgd', 'batch_size': 32, 'dropout':
0.2}
wersja_2 = {'lr': 0.001, 'epochs': 100, 'optimizer': 'adam', 'batch_size': 64, 'momentum':
0.9}

zbior_1 = set(wersja_1.keys())
zbior_2 = set(wersja_2.keys())
zbior_1_2 = zbior_1 | zbior_2
zbior_wspolny = zbior_1 & zbior_2
zbior_1_minus_2 = zbior_1 - zbior_2
zbior_2_minus_1 = zbior_2 - zbior_1

rozne_lr = wersja_1['lr'] != wersja_2['lr']
rozne_epochs = wersja_1['epochs'] != wersja_2['epochs']
rozne_optimizer = wersja_1['optimizer'] != wersja_2['optimizer']
rozne_batch_size = wersja_1['batch_size'] != wersja_2['batch_size']

print(rozne_lr)
print(rozne_epochs)
print(rozne_optimizer)
print(rozne_batch_size)

zbior_roznych_wartosci = set()
if rozne_lr:
  zbior_roznych_wartosci.add('lr')
if rozne_epochs:
  zbior_roznych_wartosci.add('epochs')
if rozne_optimizer:
  zbior_roznych_wartosci.add('optimizer')
if rozne_batch_size:
  zbior_roznych_wartosci.add('batch_size')
print(zbior_roznych_wartosci)
```
