```python
class Model:
  def __init__(self, nazwa):
    """Bazowa klasa dla wszystkich modeli ML"""
    self.nazwa = nazwa
    self.wytrenowany = False
  def trenuj(self, dane):
    """Podstawowa metoda trenowania (placeholder)"""
    print(f"Trenuję model {self.nazwa}...")
    self.wytrenowany = True
  def czy_wytrenowany(self):
    """Sprawdza, czy model jest wytrenowany"""
    return self.wytrenowany
    # Klasa potomna (dziecko) dziedziczy po Model
class ModelRegresji(Model):
  def __init__(self, nazwa, typ_regresji="liniowa"):
    """
    Konstruktor klasy potomnej.
    super().__init__(nazwa) wywołuje konstruktor klasy bazowej.
    """
    super().__init__(nazwa) # Wywołanie konstruktora rodzica
    self.typ_regresji = typ_regresji # Nowy atrybut specyficzny dla regresji
  def przewiduj(self, x):
    """Metoda specyficzna dla regresji"""
    if not self.wytrenowany:
      print("Błąd: Model nie jest wytrenowany!")
    return None
    # Prosta symulacja predykcji
    return 2.5 * x + 1.3

# Użycie
model_reg = ModelRegresji("Regresja Sprzedaży", typ_regresji="liniowa")
print(f"Czy wytrenowany: {model_reg.czy_wytrenowany()}")
model_reg.trenuj([1, 2, 3, 4, 5]) # Metoda z klasy bazowej
print(f"Czy wytrenowany: {model_reg.czy_wytrenowany()}")
predykcja = model_reg.przewiduj(10) # Metoda specyficzna dla ModelRegresji
print(f"Predykcja dla x=10: {predykcja}")
```

```python
# zadanie 9

class FiltrowanieDodatnich:
  def __init__(self):
    self.nazwa = "Filtrowanie Dodatnich Liczb"
  def akcja(self,dane):
    return [x for x in dane if x > 0]

class Zaokraglenie:
  def __init__ (self):
    self.nazwa = 'Zaokraglenie Liczb'
  def akcja(self, dane):
    return [round(x) for x in dane]

class Skalowanie:
  def __init__ (self, wspolczynnik):
    self.nazwa = 'Skalowanie Liczb'
    self.wspolczynnik = wspolczynnik
  def akcja(self, dane):
    return [x * self.wspolczynnik for x in dane]

class Sumowanie:
  def __init__ (self):
    self.nazwa = 'Sumowanie Liczb'
  def akcja(self, dane):
    return sum(dane)

#glowny pipieline
class Pipeline:
  def __init__(self, nazwa):
    self.nazwa = nazwa
    self.lista_krokow = []

  def dodaj_kroki(self,metoda):
    self.lista_krokow.append(metoda)
    return self.lista_krokow

  def uruchom(self, dane):
    wynik = dane
    for krok in self.lista_krokow:
      wynik = krok.akcja(wynik)
      print(wynik, krok.nazwa) # testowo
    return wynik

dane_1 = [1.7, -3.2, 5.8, -1.1, 8.3, 2.4]

moj_pipeline = Pipeline('Pipelie Główny')
moj_pipeline.dodaj_kroki(FiltrowanieDodatnich())
moj_pipeline.dodaj_kroki(Zaokraglenie())
moj_pipeline.dodaj_kroki(Skalowanie(3))
moj_pipeline.dodaj_kroki(Sumowanie())

moj_pipeline.uruchom(dane_1)
```

```python
# Komponenty zadanie z lekcji
class Preprocessor:
  def __init__(self):
   self.nazwa = "Preprocessor"
  def przetworz(self, dane):
   """Prostsze przetwarzanie – usunięcie None"""
   print(f" [{self.nazwa}] Przetwarzam dane...")
   return [x for x in dane if x is not None]
class Ewaluator:
  def __init__(self):
    self.nazwa = "Ewaluator"
  def ocen(self, prawdziwe, przewidywane):
    """Oblicza błąd średniokwadratowy (MSE)"""
    print(f" [{self.nazwa}] Obliczam MSE...")
    if len(prawdziwe) != len(przewidywane):
     return None
    mse = sum((p - t)**2 for p, t in zip(prawdziwe, przewidywane)) / len(prawdziwe)
    return mse
    # Klasa Model używa kompozycji – zawiera Preprocessor i Ewaluator
class ModelML:
  def __init__(self, nazwa):
    """
    Model ML składa się z preprocessora i ewaluatora.
    """
    self.nazwa = nazwa
    self.preprocessor = Preprocessor() # Kompozycja: ma preprocessor
    self.ewaluator = Ewaluator() # Kompozycja: ma ewaluator
    self.parametry = None
  def trenuj(self, dane):
    """Symulacja treningu"""
    dane_czysty = self.preprocessor.przetworz(dane)
    print(f" [Model {self.nazwa}] Trenuję na {len(dane_czysty)} obserwacjach...")
    # Prosta symulacja: parametr = średnia
    self.parametry = sum(dane_czysty) / len(dane_czysty)
    print(f" Parametr modelu: {self.parametry:.2f}")
  def przewiduj(self, x):
    """Prosta predykcja"""
    if self.parametry is None:
      print("Model nie jest wytrenowany!")
      return None
    return self.parametry * x
  def ocen_model(self, prawdziwe, przewidywane):
    """Wykorzystuje wbudowany ewaluator"""
    return self.ewaluator.ocen(prawdziwe, przewidywane)
# Użycie
model = ModelML("Regresja Prosta")
model.trenuj([10, 20, None, 30, 40, None, 50])
predykcje = [model.przewiduj(1), model.przewiduj(2), model.przewiduj(3)]
prawdziwe = [30, 60, 90]
mse = model.ocen_model(prawdziwe, predykcje)
print(f"\nMSE: {mse:.2f}")
```

```python
# zadanie od gemini:

class Utwor:
  def __init__(self,tytul,wykonawca,czas_trwania):
    self.tytul = tytul
    self.wykonawca = wykonawca
    self.czas_trwania = czas_trwania

  def __str__(self):
    return f"Utwor: {self.tytul}, Wykonawca: {self.wykonawca}, Czas trwania: {self.czas_trwania} sekund"

class Playlist:
  def __init__(self,nazwa):
    self.nazwa = nazwa
    self.utwory = []

  def dodaj_utwor(self, utwor):
    self.utwory.append(utwor)

  def calkowity_czas(self):
    czas_total = 0
    for utwor in self.utwory:
      czas_total += utwor.czas_trwania
    return czas_total

  def pokaz_playliste(self):
    print(f"Playlist name: {self.nazwa}")
    print('='*30)
    for x in self.utwory:
      print(x)
    print(f"Calkowity czas trwania: {round(self.calkowity_czas()/60)} minut")

utwor_1 = Utwor('Parostatek','Krawczyk',215)
utwor_2 = Utwor('Malgoska','Rodowicz',187)
utwor_3 = Utwor('Ale jazz','Sanah',211)
utwor_4 = Utwor('Jakis tytul','Wykonawca1',174)

trening_playlist = Playlist('Playlista treningowa')

trening_playlist.dodaj_utwor(utwor_1)
trening_playlist.dodaj_utwor(utwor_2)
trening_playlist.dodaj_utwor(utwor_3)
trening_playlist.dodaj_utwor(utwor_4)

trening_playlist.pokaz_playliste()

```

```python
# zadanie 11

class Uczen:
  def __init__(self,imie, oceny = None):
    self.imie = imie
    if oceny == None:
      self.oceny = []
    else:
      self.oceny = oceny


class Klasa:
  def __init__(self,nazwa):
    self.nazwa = nazwa
    self.uczniowie = []
  def dodaj_ucznia(self,uczen):
    self.uczniowie.append(uczen)
  def srednia_klasy(self):
    suma_ocen = 0
    liczba_ocen = 0
    for uczen in self.uczniowie:
      for ocena in uczen.oceny:
        liczba_ocen += 1
        suma_ocen += ocena
    return round((suma_ocen / liczba_ocen),2)
  def najlepszy_uczen(self):
    lista_srednich = [(uczen.imie, round(sum(uczen.oceny)/len(uczen.oceny),2)) for uczen in self.uczniowie]
    lista_srednich_malejaco = sorted(lista_srednich, key=lambda x: x[1],reverse=True)
    return lista_srednich_malejaco[0][0]


uczen_1 = Uczen('Adam',[2,3,4,5,5])
uczen_2 = Uczen('Ewa',[2,3,2,1,3,4,5,2,2])
uczen_3 = Uczen('Ania',[4,5,6,5,6,6,6,6,5])
uczen_4 = Uczen('Kasia',[2,3,4,5,5])
uczen_5 = Uczen('Darek',[1,3,4,5,5])

klasa_1c = Klasa('Klasa 1 C')

klasa_1c.dodaj_ucznia(uczen_1)
klasa_1c.dodaj_ucznia(uczen_2)
klasa_1c.dodaj_ucznia(uczen_3)
klasa_1c.dodaj_ucznia(uczen_4)
klasa_1c.dodaj_ucznia(uczen_5)

print(f"Srednia klasy: {klasa_1c.srednia_klasy()}")
klasa_1c.najlepszy_uczen()

```

```python
import math  # Potrzebne nam będzie do liczby Pi (math.pi)

# 1. Klasa bazowa (rodzic)
class Figura:
    def pole(self):
        # Klasa bazowa daje tylko "interfejs" – informację, że każda figura powinna mieć tę metodę
        raise NotImplementedError("Metoda pole() musi być zaimplementowana w klasie potomnej!")


# 2. Klasa potomna: Prostokąt
class Prostokat(Figura):
    def __init__(self, szerokosc, wysokosc):
        self.szerokosc = szerokosc
        self.wysokosc = wysokosc

    # Nadpisujemy (override) metodę pole() dla prostokąta
    def pole(self):
        return self.szerokosc * self.wysokosc


# 3. Klasa potomna: Koło
class Kolo(Figura):
    def __init__(self, promien):
        self.promien = promien

    # Nadpisujemy metodę pole() dla koła (Wzór: pi * r^2)
    def pole(self):
        return math.pi * (self.promien ** 2)


# --- Testowanie systemu i liczenie łącznego pola ---

# Tworzymy listę różnych figur
figury = [
    Prostokat(4, 5),   # Pole = 20
    Kolo(3),           # Pole = ~28.27 (pi * 3^2)
    Prostokat(2, 10),  # Pole = 20
    Kolo(2)            # Pole = ~12.57 (pi * 2^2)
]

# Obliczamy łączne pole przy użyciu funkcji sum() i generatora
# Zauważ magię polimorfizmu: dla każdego elementu wywołujemy po prostu .pole(),
# a Python sam wie, czy ma wykonać wzór prostokąta, czy koła!
calkowite_pole = sum(figura.pole() for figura in figury)

print(f"Łączne pole wszystkich figur wynosi: {calkowite_pole:.2f}")
```

```python
# zadanie 14 - GEMINI KOD

import math

# 1. TRANSFORMER 1: Przekształcenie logarytmiczne log(x + 1)
class LogTransformer:
    def transform(self, data):
        # Dla każdego elementu liczymy logarytm naturalny z (x + 1)
        return [math.log(x + 1) for x in data]


# 2. TRANSFORMER 2: Usuwanie wartości odstających (Outliers) > 3 * std od średniej
class OutlierRemover:
    def transform(self, data):
        if not data:
            return []

        # Krok A: Liczymy średnią
        mean = sum(data) / len(data)

        # Krok B: Liczymy odchylenie standardowe (std)
        variance = sum((x - mean) ** 2 for x in data) / len(data)
        std = math.sqrt(variance)

        # Krok C: Ustalamy próg (3 * std od średniej w obie strony)
        threshold = 3 * std

        # Krok D: Zostawiamy tylko te elementy, które mieszczą się w granicach
        filtered_data = [x for x in data if abs(x - mean) <= threshold]
        return filtered_data


# 3. TRANSFORMER 3: Dzielenie na przedziały (Binning)
class BinningTransformer:
    def __init__(self, num_bins=3):
        self.num_bins = num_bins  # Liczba przedziałów

    def transform(self, data):
        if not data:
            return []

        min_val = min(data)
        max_val = max(data)

        # Jeśli wszystkie wartości są takie same
        if min_val == max_val:
            return [0 for _ in data]

        # Szerokość pojedynczego przedziału
        bin_width = (max_val - min_val) / self.num_bins

        binned_data = []
        for x in data:
            # Obliczamy numer przedziału dla danej wartości
            bin_idx = int((x - min_val) / bin_width)
            # Zabezpieczenie dla wartości maksymalnej (trafi do ostatniego koszyka)
            if bin_idx >= self.num_bins:
                bin_idx = self.num_bins - 1
            binned_data.append(bin_idx)

        return binned_data


# 4. KLASA PIPELINE: Łączy transformery w jeden ciąg operacji
class Pipeline:
    def __init__(self, steps):
        self.steps = steps  # Lista kolejnych transformerów do wykonania

    def fit_transform(self, data):
        current_data = data
        # Przepuszczamy dane przez każdy krok po kolei
        for step in self.steps:
            current_data = step.transform(current_data)
            # Pobieramy nazwę klasy obiektu 'step' w locie:
            nazwa_klasy = type(step).__name__
            print(f"Po kroku [{nazwa_klasy}]: {current_data}")
        return current_data

# Dane testowe
dane_wejsciowe = [1, 5, 10, 100, 2, 3, 8, 200, 4, 6]

print(f"Dane początkowe: {dane_wejsciowe}\n")

# Tworzymy pipeline składający się z kolejnych kroków
moj_pipeline = Pipeline([
    OutlierRemover(),     # 1. Najpierw wyrzucamy wartości odstające (100 i 200)
    LogTransformer(),     # 2. Następnie aplikujemy log(x + 1) na oczyszczone dane
    BinningTransformer(num_bins=3)  # 3. Na koniec dzielimy wyniki na 3 przedziały
])

# Uruchamiamy cały proces
wynik = moj_pipeline.fit_transform(dane_wejsciowe)

print(f"Wynik końcowy po przejściu przez Pipeline: {wynik}")
```

```python
# zadanie 15 GEMINI KOD DO STUDIOWANIA
# ==========================================
# KLASA REPREZENTUJĄCĄ POJEDYNCZĄ AKCJĘ
# ==========================================
class Akcja:
    def __init__(self, nazwa, cena_kupna, cena_aktualna=None):
        """
        Konstruktor klasy Akcja.
        :param nazwa: nazwa spółki / symbol (np. 'AAPL', 'ORLEN')
        :param cena_kupna: cena, po której pierwotnie kupiono akcję
        :param cena_aktualna: bieżąca cena rynkowa (jeśli None, domyślnie przyjmuje cenę zakupu)
        """
        self.nazwa = nazwa
        self.cena_kupna = cena_kupna
        # Jeśli nie podano ceny aktualnej przy tworzeniu, ustawiamy ją na cenę zakupu
        self.cena_aktualna = cena_aktualna if cena_aktualna is not None else cena_kupna

    def __repr__(self):
        # Pomocnicza metoda do ładnego wyświetlania obiektu w konsoli
        return f"Akcja({self.nazwa}, kupno: {self.cena_kupna} PLN, aktualna: {self.cena_aktualna} PLN)"


# ==========================================
# KLASA REPREZENTUJĄCĄ PORTFEL INWESTYCYJNY
# ==========================================
class Portfel:
    def __init__(self, kapital_poczatkowy):
        """
        Konstruktor klasy Portfel.
        :param kapital_poczatkowy: początkowa ilość gotówki przeznaczona na inwestycje
        """
        self.kapital_poczatkowy = kapital_poczatkowy
        self.gotowka = kapital_poczatkowy  # Gotówka pomniejszana przy zakupach i powiększana przy sprzedaży
        self.akcje = {}  # Słownik przechowujący akcje w portfelu w formacie: { obiekt_akcji: ilosc }

    def kup_akcje(self, akcja, ilosc):
        """
        Metoda obsługująca kupno akcji.
        """
        koszt_calkowity = akcja.cena_aktualna * ilosc

        # Sprawdzamy, czy inwestor ma wystarczająco dużo gotówki
        if self.gotowka >= koszt_calkowity:
            self.gotowka -= koszt_calkowity  # Odejmujemy koszt zakupu od dostępnej gotówki

            # Jeśli akcja już jest w portfelu, zwiększamy jej ilość, w przeciwnym razie dodajemy nową pozycję
            if akcja in self.akcje:
                self.akcje[akcja] += ilosc
            else:
                self.akcje[akcja] = ilosc

            print(f"[KUPNO] Kupiono {ilosc} szt. akcji {akcja.nazwa} za łączną kwotę {koszt_calkowity:.2f} PLN.")
        else:
            print(f"[BŁĄD] Brak wystarczających środków na zakup {ilosc} szt. {akcja.nazwa}! Wymagane: {koszt_calkowity:.2f} PLN, masz: {self.gotowka:.2f} PLN.")

    def sprzedaj_akcje(self, akcja, ilosc):
        """
        Metoda obsługująca sprzedaż akcji.
        """
        # Sprawdzamy, czy akcja w ogóle znajduje się w portfelu i czy mamy jej wystarczającą ilość
        if akcja in self.akcje and self.akcje[akcja] >= ilosc:
            wplyw_ze_sprzedazy = akcja.cena_aktualna * ilosc
            self.gotowka += wplyw_ze_sprzedazy  # Dodajemy gotówkę ze sprzedaży

            self.akcje[akcja] -= ilosc  # Zmniejszamy ilość akcji w portfelu

            # Jeśli ilość spadła do zera, całkowicie usuwamy akcję ze słownika portfela
            if self.akcje[akcja] == 0:
                del self.akcje[akcja]

            print(f"[SPRZEDAŻ] Sprzedano {ilosc} szt. akcji {akcja.nazwa} za łączną kwotę {wplyw_ze_sprzedazy:.2f} PLN.")
        else:
            posiadana_ilosc = self.akcje.get(akcja, 0)
            print(f"[BŁĄD] Nie możesz sprzedać {ilosc} szt. {akcja.nazwa}. Posiadasz tylko {posiadana_ilosc} szt.")

    def aktualizuj_ceny(self, slownik_cen):
        """
        Aktualizuje bieżące ceny akcji na podstawie przekazanego słownika.
        :param slownik_cen: słownik w formacie { 'NazwaAkcji': nowa_cena }
        """
        print("\n--- AKTUALIZACJA CEN NA RYNKU ---")
        # Przechodzimy przez wszystkie akcje, które aktualnie mamy w portfelu
        for akcja in self.akcje.keys():
            if akcja.nazwa in slownik_cen:
                stara_cena = akcja.cena_aktualna
                akcja.cena_aktualna = slownik_cen[akcja.nazwa]
                print(f"Cena {akcja.nazwa} zaktualizowana: {stara_cena} PLN -> {akcja.cena_aktualna} PLN")

    def wartosc_portfela(self):
        """
        Oblicza łączną wartość portfela (gotówka + rynkowa wartość posiadanych akcji).
        """
        wartosc_akcji = 0
        for akcja, ilosc in self.akcje.items():
            wartosc_akcji += akcja.cena_aktualna * ilosc

        return self.gotowka + wartosc_akcji

    def zysk_strata(self):
        """
        Oblicza różnicę (wartość obecna portfela minus kapitał początkowy).
        Zwraca kwotę zysku (dodatnia) lub straty (ujemna).
        """
        return self.wartosc_portfela() - self.kapital_poczatkowy

    def status_portfela(self):
        """
        Wyświetla podsumowanie stanu portfela.
        """
        print("\n" + "="*40)
        print(" STAN PORTFELA ")
        print("="*40)
        print(f"Kapitał początkowy: {self.kapital_poczatkowy:.2f} PLN")
        print(f"Dostępna gotówka:   {self.gotowka:.2f} PLN")
        print("Posiadane akcje:")
        if not self.akcje:
            print("  Brak akcji w portfelu.")
        else:
            for akcja, ilosc in self.akcje.items():
                wartosc_pozycji = akcja.cena_aktualna * ilosc
                print(f"  - {akcja.nazwa}: {ilosc} szt. | Cena akt.: {akcja.cena_aktualna} PLN | Wartość: {wartosc_pozycji:.2f} PLN")

        akt_wartosc = self.wartosc_portfela()
        wynik = self.zysk_strata()
        print(("-" * 40))
        print(f"Łączna wartość portfela: {akt_wartosc:.2f} PLN")
        print(f"Łączny zysk/strata:      {wynik:+.2f} PLN")
        print("="*40 + "\n")


# ==========================================
# SYMULACJA DZIAŁANIA PORTFELA
# ==========================================
if __name__ == "__main__":
    print("Rozpoczynamy symulację portfela inwestycyjnego...\n")

    # 1. Tworzymy 5 różnych akcji z cenami początkowymi
    akcja1 = Akcja("ORLEN", 60.0)
    akcja2 = Akcja("KGHM", 140.0)
    akcja3 = Akcja("CD_PROJEKT", 120.0)
    akcja4 = Akcja("ALLEGRO", 35.0)
    akcja5 = Akcja("PKO_BP", 45.0)

    # 2. Inicjalizujemy portfel z kapitałem początkowym 10 000 PLN
    moj_portfel = Portfel(kapital_poczatkowy=10000.0)
    moj_portfel.status_portfela()

    # 3. Kupujemy różne ilości akcji
    moj_portfel.kup_akcje(akcja1, 30)  # ORLEN: 30 * 60 = 1800 PLN
    moj_portfel.kup_akcje(akcja2, 10)  # KGHM: 10 * 140 = 1400 PLN
    moj_portfel.kup_akcje(akcja3, 20)  # CD_PROJEKT: 20 * 120 = 2400 PLN
    moj_portfel.kup_akcje(akcja4, 50)  # ALLEGRO: 50 * 35 = 1750 PLN
    moj_portfel.kup_akcje(akcja5, 40)  # PKO_BP: 40 * 45 = 1800 PLN
    # Łącznie wydano: 1800 + 1400 + 2400 + 1750 + 1800 = 9150 PLN. Zostaje 850 PLN gotówki.

    moj_portfel.status_portfela()

    # 4. Symulujemy zmianę cen na rynku (np. publikacja raportów kwartalnych)
    nowe_ceny_rynkowe = {
        "ORLEN": 68.5,       # Wzrost
        "KGHM": 130.0,       # Spadek
        "CD_PROJEKT": 135.0, # Wzrost
        "ALLEGRO": 32.0,     # Spadek
        "PKO_BP": 50.0       # Wzrost
    }

    moj_portfel.aktualizuj_ceny(nowe_ceny_rynkowe)
    moj_portfel.status_portfela()

    # 5. Dokonujemy transakcji sprzedaży części akcji (realizacja zysków/strat)
    print("--- TRANSAKCJE SPRZEDAŻY ---")
    moj_portfel.sprzedaj_akcje(akcja1, 10) # Sprzedaż 10 szt. ORLEN po nowej cenie 68.5
    moj_portfel.sprzedaj_akcje(akcja4, 50) # Sprzedaż wszystkich 50 szt. ALLEGRO po nowej cenie 32.0

    # 6. Ostateczny stan portfela
    moj_portfel.status_portfela()
```

```python
# zadanie 17 KOD GEMINI - do edukacji.

from datetime import datetime

# ==========================================
# 1. KLASA REPREZENTUJĄCA POJEDYNCZY POKÓJ
# ==========================================
class Pokoj:
    def __init__(self, numer, typ, cena_za_noc, zajety=False):
        """
        Konstruktor klasy Pokoj.
        :param numer: unikalny numer pokoju (np. 101, 202)
        :param typ: rodzaj pokoju (np. 'Jednoosobowy', 'Apartament')
        :param cena_za_noc: cena za jedną dobę hotelową w PLN
        :param zajety: status pokoju (domyślnie False – wolny)
        """
        self.numer = numer
        self.typ = typ
        self.cena_za_noc = cena_za_noc
        self.zajety = zajety

    def __repr__(self):
        status = "Zajęty" if self.zajety else "Wolny"
        return f"Pokój {self.numer} ({self.typ}) - {self.cena_za_noc} PLN/noc [{status}]"


# ==========================================
# 2. KLASA REPREZENTUJĄCA REZERWACJĘ
# ==========================================
class Rezerwacja:
    def __init__(self, pokoj, data_od, data_do, gosc):
        """
        Konstruktor klasy Rezerwacja.
        :param pokoj: obiekt klasy Pokoj, którego dotyczy rezerwacja
        :param data_od: data zameldowania (jako napis w formacie 'YYYY-MM-DD')
        :param data_do: data wymeldowania (jako napis w formacie 'YYYY-MM-DD')
        :param gosc: imię i nazwisko gościa
        """
        self.pokoj = pokoj
        self.gosc = gosc

        # Konwertujemy napisy (stringi) na obiekty dat (datetime),
        # dzięki czemu będziemy mogli łatwo obliczyć liczbę dni pobytu.
        self.data_od = datetime.strptime(data_od, "%Y-%m-%d")
        self.data_do = datetime.strptime(data_do, "%Y-%m-%d")

    def oblicz_koszt(self):
        """
        Oblicza całkowity koszt rezerwacji (liczba dni * cena za noc).
        """
        # Różnica dat daje obiekt timedelta, z którego wyciągamy liczbę dni (.days)
        liczba_dni = (self.data_do - self.data_od).days

        # Zabezpieczenie na wypadek błędnego podania dat (np. min. 1 doba)
        if liczba_dni <= 0:
            liczba_dni = 1

        return liczba_dni * self.pokoj.cena_za_noc

    def __repr__(self):
        koszt = self.oblicz_koszt()
        return (f"Rezerwacja: Pokój {self.pokoj.numer} dla {self.gosc} "
                f"({self.data_od.strftime('%Y-%m-%d')} do {self.data_do.strftime('%Y-%m-%d')}) "
                f"| Koszt: {koszt} PLN")


# ==========================================
# 3. KLASA ZARZĄDZAJĄCA HOTELEM
# ==========================================
class Hotel:
    def __init__(self):
        """
        Konstruktor klasy Hotel. Inicjalizuje puste listy na pokoje i rezerwacje.
        """
        self.pokoje = []
        self.rezerwacje = []

    def dodaj_pokoj(self, pokoj):
        """Dodaje nowy pokój do bazy hotelu."""
        self.pokoje.append(pokoj)

    def rezerwuj(self, numer_pokoju, gosc, data_od, data_do):
        """
        Metoda obsługująca proces rezerwacji pokoju.
        """
        # Szukamy pokoju o podanym numerze w naszej liście pokoi
        docelowy_pokoj = None
        for p in self.pokoje:
            if p.numer == numer_pokoju:
                docelowy_pokoj = p
                break  # Przerywamy pętlę, gdy znaleźliśmy pokój

        # Sprawdzenie 1: Czy pokój w ogóle istnieje w hotelu?
        if docelowy_pokoj is None:
            print(f"[BŁĄD] Pokój numer {numer_pokoju} nie istnieje w hotelu!")
            return

        # Sprawdzenie 2: Czy pokój jest aktualnie wolny?
        if docelowy_pokoj.zajety:
            print(f"[BŁĄD] Pokój {numer_pokoju} jest już zajęty!")
            return

        # Jeśli wszystko jest OK, tworzymy rezerwację
        nowa_rezerwacja = Rezerwacja(docelowy_pokoj, data_od, data_do, gosc)

        # Zapisujemy rezerwację na liście hotelu
        self.rezerwacje.append(nowa_rezerwacja)

        # Zmieniamy status pokoju na zajęty
        docelowy_pokoj.zajety = True

        print(f"[SUKCES] Zarezerwowano pokój {numer_pokoju} dla {gosc}. "
              f"Do zapłaty: {nowa_rezerwacja.oblicz_koszt():.2f} PLN")

    def wolne_pokoje(self):
        """
        Zwraca listę pokoi, które mają status zajety = False.
        """
        wolne = []
        for p in self.pokoje:
            if not p.zajety:
                wolne.append(p)
        return wolne

    def przychod(self):
        """
        Oblicza sumę przychodów ze wszystkich zrealizowanych rezerwacji.
        """
        calkowity_przychod = 0
        for rez in self.rezerwacje:
            calkowity_przychod += rez.oblicz_koszt()
        return calkowity_przychod


# ==========================================
# SYMULACJA DZIAŁANIA SYSTEMU HOTELOWEGO
# ==========================================
if __name__ == "__main__":
    print("--- INICJALIZACJA HOTELU ---")
    mój_hotel = Hotel()

    # 1. Tworzymy 10 różnych pokoi i dodajemy je do hotelu
    mój_hotel.dodaj_pokoj(Pokoj(101, "Jednoosobowy", 150.0))
    mój_hotel.dodaj_pokoj(Pokoj(102, "Jednoosobowy", 150.0))
    mój_hotel.dodaj_pokoj(Pokoj(201, "Dwuosobowy", 250.0))
    mój_hotel.dodaj_pokoj(Pokoj(202, "Dwuosobowy", 250.0))
    mój_hotel.dodaj_pokoj(Pokoj(203, "Dwuosobowy", 270.0))
    mój_hotel.dodaj_pokoj(Pokoj(301, "Apartament", 500.0))
    mój_hotel.dodaj_pokoj(Pokoj(302, "Apartament", 550.0))
    mój_hotel.dodaj_pokoj(Pokoj(401, "Studio", 350.0))
    mój_hotel.dodaj_pokoj(Pokoj(402, "Studio", 350.0))
    mój_hotel.dodaj_pokoj(Pokoj(501, "Prezydencki", 1200.0))

    print(f"Dodano łącznie {len(mój_hotel.pokoje)} pokoi do hotelu.\n")

    # 2. Sprawdzamy stan wolnych pokoi przed rezerwacjami
    print(f"Liczba wolnych pokoi na start: {len(mój_hotel.wolne_pokoje())}")

    print("\n--- DOKONUJEMY 5 REZERWACJI ---")
    # 3. Symulujemy 5 rezerwacji
    mój_hotel.rezerwuj(101, "Jan Kowalski", "2026-07-01", "2026-07-05")     # 4 noce * 150 = 600 PLN
    mój_hotel.rezerwuj(201, "Anna Nowak", "2026-07-02", "2026-07-07")       # 5 nocy * 250 = 1250 PLN
    mój_hotel.rezerwuj(301, "Elon Musk", "2026-07-10", "2026-07-12")        # 2 noce * 500 = 1000 PLN
    mój_hotel.rezerwuj(401, "Piotr Zieliński", "2026-08-01", "2026-08-08")  # 7 nocy * 350 = 2450 PLN
    mój_hotel.rezerwuj(501, "Robert Lewandowski", "2026-06-15", "2026-06-20") # 5 nocy * 1200 = 6000 PLN

    # Próba zarezerwowania pokoju, który jest już zajęty (test zabezpieczenia)
    print("\nPróba ponownej rezerwacji zajętego pokoju:")
    mój_hotel.rezerwuj(101, "Ktoś Inny", "2026-07-02", "2026-07-03")

    print("\n--- PODSUMOWANIE HOTELU ---")
    # 4. Sprawdzamy liczbę wolnych pokoi po rezerwacjach
    wolne = mój_hotel.wolne_pokoje()
    print(f"Liczba aktualnie wolnych pokoi: {len(wolne)}")

    # 5. Sprawdzamy całkowity przychód hotelu
    calkowity_przychod = mój_hotel.przychod()
    print(f"Całkowity przychód ze wszystkich rezerwacji: {calkowity_przychod:.2f} PLN")
```

```python
# zadanie 18 - kod 100% GEMINI - przestudiowany
import math


class ValidatorZakresu:
  """Sprawdza, czy wszystkie wartości mieszczą się w podanym zakresie [min, max]."""

  def __init__(self, min_val, max_val):
    self.min_val = min_val
    self.max_val = max_val

  def validate(self, data):
    for x in data:
      # Ignorujemy None/NaN w tym walidatorze, jeśli obsługuje je ValidatorBrakow
      if x is not None and not (isinstance(x, (int, float)) and math.isnan(x)):
        if not (self.min_val <= x <= self.max_val):
          return (
              False,
              f"Wartość {x} jest poza zakresem [{self.min_val}," f" {self.max_val}].",)
    return True, "Zakres jest poprawny."


class ValidatorTypu:
  """Sprawdza, czy wszystkie wartości są określonego typu."""

  def __init__(self, typ):
    self.typ = typ

  def validate(self, data):
    for x in data:
      if not isinstance(x, self.typ):
        return (
            False,
            f"Wartość '{x}' ma typ {type(x).__name__}, a oczekiwano"
            f" {self.typ.__name__}.",
        )
    return True, "Wszystkie typy są poprawne."


class ValidatorBrakow:
  """Sprawdza, czy w danych nie występują wartości None lub NaN."""

  def validate(self, data):
    for x in data:
      if x is None:
        return False, "Wykryto wartość None."
      if isinstance(x, float) and math.isnan(x):
        return False, "Wykryto wartość NaN."
    return True, "Brak braków danych (None/NaN)."


class Pipeline:
  """Pipeline agregujący i uruchamiający wiele walidatorów."""

  def __init__(self):
    self.validators = []

  def dodaj(self, validator):
    self.validators.append(validator)
    return self  # Umożliwia łączenie metod w łańcuch

  def uruchom(self, data):
    wyniki = []
    for validator in self.validators:
      status, komunikat = validator.validate(data)
      wyniki.append({
          "validator": validator.__class__.__name__,
          "sukces": status,
          "komunikat": komunikat,
      })
    return wyniki

####################
# Przykładowe dane do testów
dane_poprawne = [10, 20, 30, 45, 50]
dane_z_bledami = [10, None, 150, "tekst", float("nan")]

# Tworzenie pipeline'u
pipeline = (
    Pipeline()
    .dodaj(ValidatorBrakow())
    .dodaj(ValidatorTypu(int))
    .dodaj(ValidatorZakresu(0, 100))
)

print("--- Test 1: Dane poprawne ---")
wyniki_1 = pipeline.uruchom(dane_poprawne)
for w in wyniki_1:
  print(f"{w['validator']}: {w['sukces']} - {w['komunikat']}")

print("\n--- Test 2: Dane z błędami ---")
wyniki_2 = pipeline.uruchom(dane_z_bledami)
for w in wyniki_2:
  print(f"{w['validator']}: {w['sukces']} - {w['komunikat']}")

```

```python
# zadanie 19 - kod 100% gemini przestudiowany

import random


class Postac:
  """Klasa bazowa reprezentująca postać w grze RPG."""

  def __init__(self, imie, hp, atak, obrona):
    self.imie = imie
    self.max_hp = hp
    self.hp = hp
    self.atak = atak
    self.obrona = obrona

  def czy_zyje(self):
    return self.hp > 0

  def atakuj(self, cel):
    # Obliczanie obrażeń: atak minus obrona celu (minimum 1 punkt obrażeń)
    obrazenia = max(1, self.atak - cel.obrona)
    cel.hp = max(0, cel.hp - obrazenia)
    print(
        f"⚔️ {self.imie} atakuje {cel.imie} i zadaje {obrazenia} pkt"
        f" obrażeń! ({cel.imie} ma teraz {cel.hp}/{cel.max_hp} HP)"
    )

  def leczenie(self, punkty):
    self.hp = min(self.max_hp, self.hp + punkty)
    print(
        f"🧪 {self.imie} leczy się o {punkty} HP. Aktualne HP: {self.hp}"
        f"/{self.max_hp}"
    )

  def status(self):
    return f"{self.imie} [HP: {self.hp}/{self.max_hp}, Atak: {self.atak}, Obrona: {self.obrona}]"


class Wojownik(Postac):
  """Wojownik – posiada zwiększony atak oraz potężny cios krytyczny."""

  def __init__(self, imie, hp=120, atak=25, obrona=10):
    super().__init__(imie, hp, atak, obrona)
    self.klasa = "Wojownik"

  def atakuj(self, cel):
    # Wojownik ma 30% szans na potężny cios (2x obrażenia)
    czy_kryt = random.random() < 0.3
    if czy_kryt:
      obrazenia = max(1, (self.atak * 2) - cel.obrona)
      cel.hp = max(0, cel.hp - obrazenia)
      print(
          f"💥 [KRYTYK!] {self.imie} zadaje potężny cios mieczem celowi"
          f" {cel.imie}! Zadano {obrazenia} pkt obrażeń! ({cel.imie} ma"
          f" {cel.hp}/{cel.max_hp} HP)"
      )
    else:
      super().atakuj(cel)


class Mag(Postac):
  """Mag – potrafi rzucać potężne zaklęcia ignorujące obronę."""

  def __init__(self, imie, hp=80, atak=15, obrona=4, mana=50):
    super().__init__(imie, hp, atak, obrona)
    self.mana = mana
    self.klasa = "Mag"

  def rzuc_czar(self, cel):
    if self.mana >= 20:
      self.mana -= 20
      # Czary ignorują obronę przeciwnika
      obrazenia = 35
      cel.hp = max(0, cel.hp - obrazenia)
      print(
          f"🔮 {self.imie} rzuca Kule Ognia w {cel.imie}! Ignoruje obronę i"
          f" zadaje {obrazenia} pkt obrażeń! ({cel.imie} ma"
          f" {cel.hp}/{cel.max_hp} HP) [Mana: {self.mana}]"
      )
    else:
      print(f"💤 {self.imie} probuje rzucić czar, ale nie ma mamy! Atakuje zwykle.")
      super().atakuj(cel)


class Lucznik(Postac):
  """Łucznik – wysoka szybkość, szansa na podwójny strzał."""

  def __init__(self, imie, hp=90, atak=20, obrona=6):
    super().__init__(imie, hp, atak, obrona)
    self.klasa = "Łucznik"

  def atakuj(self, cel):
    # Łucznik ma 40% szans na podwójny strzał
    super().atakuj(cel)
    if cel.czy_zyje() and random.random() < 0.4:
      print(
          f"🎯 [SZYBKI STRZAŁ!] {self.imie} wypuszcza drugą strzałkę"
          f" błyskawicznie!"
      )
      super().atakuj(cel)


# --- Symulacja walki ---
def symuluj_walke(postac1, postac2):
  print(f"=== ROZPOCZYNA SIĘ WALKÁ ===")
  print(f"🤼 {postac1.status()}  VS  {postac2.status()}")
  print("=" * 35)

  tura = 1
  while postac1.czy_zyje() and postac2.czy_zyje():
    print(f"\n--- Tura {tura} ---")

    # Postac 1 atakuje Postac 2
    if isinstance(postac1, Mag) and random.random() < 0.5:
      postac1.rzuc_czar(postac2)
    else:
      postac1.atakuj(postac2)

    if not postac2.czy_zyje():
      print(f"\n🏆 {postac1.imie} wygrywa walkę!")
      break

    # Postac 2 atakuje Postac 1
    if isinstance(postac2, Mag) and random.random() < 0.5:
      postac2.rzuc_czar(postac1)
    else:
      postac2.atakuj(postac1)

    if not postac1.czy_zyje():
      print(f"\n🏆 {postac2.imie} wygrywa walkę!")
      break

    tura += 1


# Uruchomienie symulacji (Wojownik vs Mag)
if __name__ == "__main__":
  wojownik = Wojownik("Conan")
  mag = Mag(
      "Gandalf"
  )  # Posiada zdolność rzucania czarów ignorujących pancerz

  symuluj_walke(wojownik, mag)
```

```python
# zadanie 20 - KOD GEMINI 100% - przestudiowane:

# ==========================================
# 1. KLASA BAZOWA (Polimorfizm i dziedziczenie)
# ==========================================

class BaseTransformer:
    """
    Klasa bazowa (rodzic) dla wszystkich transformerów.
    Definiuje wspólny interfejs, którego będą używać wszystkie podklasy.
    Dzięki temu Pipeline może traktować każdy transformer w ten sam sposób.
    """

    def fit(self, dane):
        """
        Metoda fit (dopasowanie do danych). W prostych transformerach
        zazwyczaj nic nie robi, ale w scikit-learn służy do zbierania statystyk
        (np. średniej, minimalnej/maksymalnej wartości). Zwraca self (wzorzec Builder).
        """
        return self

    def transform(self, dane):
        """
        Metoda transformująca dane. Musi zostać zaimplementowana w każdej klasie potomnej.
        """
        raise NotImplementedError("Podklasa musi zaimplementować metodę transform!")

    def fit_transform(self, dane):
        """
        Metoda łącząca fit oraz transform – analogicznie jak w scikit-learn.
        Najpierw dopasowuje transformer, a następnie przetwarza dane.
        """
        return self.fit(dane).transform(dane)


# ==========================================
# 2. KONKRETNE TRANSFORMATORY
# ==========================================

class DropNullTransformer(BaseTransformer):
    """
    Transformer usuwający wartości None z listy danych.
    """

    def transform(self, dane):
        # Używamy list comprehension, aby zostawić tylko te elementy, które nie są None
        print([x for x in dane if x is not None], "<- [DropNullTransformer]")
        return [x for x in dane if x is not None]


class ClipTransformer(BaseTransformer):
    """
    Transformer ograniczający wartości liczbowe do zadanego przedziału [min, max].
    """

    def __init__(self, min_val, max_val):
        # Konstruktor przyjmuje parametry konfigurujące zakres
        self.min_val = min_val
        self.max_val = max_val

    def transform(self, dane):
        wynik = []
        for x in dane:
            # Jeśli wartość jest mniejsza niż min, ustawiamy min; jeśli większa niż max, to max
            if x < self.min_val:
                wynik.append(self.min_val)
            elif x > self.max_val:
                wynik.append(self.max_val)
            else:
                wynik.append(x)
        print(wynik, "<- [ClipTransformer]")
        return wynik


class SquareTransformer(BaseTransformer):
    """
    Transformer podnoszący każdą wartość liczbową do kwadratu (x^2).
    """

    def transform(self, dane):
        # Podnosimy każdy element do potęgi 2
        wynik = [x ** 2 for x in dane]
        print(wynik, "<- [SquareTransformer]")
        return wynik


class InverseTransformer(BaseTransformer):
    """
    Transformer obliczający odwrotność liczby (1/x).
    """

    def transform(self, dane):
        wynik = []
        for x in dane:
            if x == 0:
                raise ValueError("Błąd: Próba dzielenia przez zero w InverseTransformer!")
            wynik.append(1 / x)
        print(wynik, "<- [InverseTransformer]")
        return wynik


# ==========================================
# 3. PIPELINE (Sekwencyjne przetwarzanie)
# ==========================================

class Pipeline:
    """
    Klasa Pipeline odpowiada za łączenie wielu transformerów w jeden ciąg operacji.
    Dane przechodzą kolejno przez każdy transformer w liście.
    """

    def __init__(self, transformers):
        # Przyjmuje listę instancji transformerów w kolejności wykonywania
        self.transformers = transformers

    def fit(self, dane):
        # Wywołuje metodę fit dla każdego transformera w potoku
        aktualne_dane = dane
        for t in self.transformers:
            t.fit(aktualne_dane)
        return self

    def transform(self, dane):
        # Przepuszcza dane przez każdy transformer po kolei (wyjście jednego to wejście następnego)
        aktualne_dane = dane
        for t in self.transformers:
            aktualne_dane = t.transform(aktualne_dane)
        return aktualne_dane

    def fit_transform(self, dane):
        """
        Zgodnie z wymaganiem: metoda fit_transform działająca analogicznie do scikit-learn.
        Wywołuje najpierw fit, a potem transform dla całego potoku.
        """
        self.fit(dane)
        return self.transform(dane)


# ==========================================
# 4. TESTOWANIE ROZWIĄZANIA
# ==========================================

if __name__ == "__main__":
    # Dane początkowe zawierające wartości None, liczby ujemne oraz wartości wykraczające poza zakres
    dane_wejsciowe = [None, -5, 1, 3, 5, 12]

    print("Dane początkowe:", dane_wejsciowe)
    print("-" * 50)

    # Tworzymy listę 4 transformerów z odpowiednimi parametrami
    # (Zakres w ClipTransformer ustawiamy od 1 do 10, aby uniknąć dzielenia przez zero w InverseTransformer)
    kroki_pipeline = [
        DropNullTransformer(),
        ClipTransformer(min_val=1, max_val=10),
        SquareTransformer(),
        InverseTransformer()
    ]

    # Inicjalizacja Pipeline
    moj_pipeline = Pipeline(kroki_pipeline)

    # Uruchomienie metody fit_transform na całym potoku
    print("Rozpoczęcie przetwarzania w Pipeline (krok po kroku):")
    wynik_koncowy = moj_pipeline.fit_transform(dane_wejsciowe)

    print("-" * 50)
    print("Ostateczny wynik po przejściu przez cały Pipeline:", wynik_koncowy)
```
