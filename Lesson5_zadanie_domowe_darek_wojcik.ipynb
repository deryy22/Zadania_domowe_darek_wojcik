```python
# zadanie 9
cm = [
      [80,10], # TN , FP
      [5,45]   # FN, TP
]

tn = cm[0][0] # true negative
fp = cm[0][1] # false positive
fn = cm[1][0] # false negative
tp = cm[1][1] # true positive


accuracy = (tp+tn) / (tp+tn+fp+fn)
precision = tp / (tp + fp)
recall = tp / (tp + fn)

print(f"Accuracy: {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
```
