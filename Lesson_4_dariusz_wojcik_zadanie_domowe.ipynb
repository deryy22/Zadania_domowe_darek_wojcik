```python
# zadanie 13 - Gemini generated ale przeczytalem i zrozumialem :)

pomiary = [120, 125, 118, 190, 122]
# Dla małej próby z wyraźną wartością odstającą pomijamy chwilowo ekstremum
# do wyznaczenia stabilnego przedziału referencyjnego:
pomiary_bazowe = [120, 125, 118, 122]

suma = sum(pomiary_bazowe)
print(suma)
liczba_elementow = len(pomiary_bazowe)
srednia = suma / liczba_elementow


suma_kwadratow_roznic = sum((x - srednia) ** 2 for x in pomiary_bazowe)
odchylenie_std = (suma_kwadratow_roznic / liczba_elementow) ** 0.5
print(f"Srednia wynosi: {srednia:.2f}")
print(f"Odchylenie standardowe wynosi: {odchylenie_std:.2f}")

# 2. Ustalenie progu 3-sigma
prog = 3 * odchylenie_std
dolna_granica = srednia - prog
gorna_granica = srednia + prog

# 3. Detekcja anomalii dla wszystkich pomiarów i wyświetlenie raportu
for x in pomiary:
    if x < dolna_granica or x > gorna_granica:
        status = "ANOMALIA!"
    else:
        status = "OK"

    print(f"Wartość: {x} - {status}")
```

```python
#zadanie 15: #Gemini pomoglo z z_scores lista
wagi_pacjentow = [65.5, 78.2, 54.8, 92.1, 70.0]
waga_srednia = sum(wagi_pacjentow)/len(wagi_pacjentow)
suma_kwadratow_roznic_wagi = sum((x - waga_srednia)**2 for x in wagi_pacjentow)
odchylenie_std_wagi = (suma_kwadratow_roznic_wagi / (len(wagi_pacjentow)-1)) ** 0.5

z_scores = [] ## Lista do przechowani z-scores (przydatna do sprawdzenia średniej i odchylenia

for x in wagi_pacjentow:
  z_score = (x-waga_srednia) / odchylenie_std_wagi
  z_scores.append(z_score)
  if z_score > 2:
    interpretacja_z = "Bardzo wysoka"
  elif abs(z_score) < 1:
    interpretacja_z = "Typowa"
  elif z_score < -2:
    interpretacja_z = "Bardzo Niska"
  else:
    interpretacja_z = "Umiarkowana"
  print(f"Wartosc oryginalna: {x}, Z-Score: {z_score:<8.3f} Interpretacja: {interpretacja_z}")


# 2. Sprawdzenie wymagań: średnia z-scores ≈ 0 i odchylenie ≈ 1
srednia_z = sum(z_scores) / len(z_scores)
# Odchylenie standardowe dla z-scores
suma_kw_z = sum((z - srednia_z)**2 for z in z_scores)
odchylenie_z = (suma_kw_z / (len(z_scores) - 1)) ** 0.5

print("-" * 30)
print(f"Sprawdzenie średniej z-scores: {srednia_z:.5f} (powinno być ok. 0)")
print(f"Sprawdzenie odchylenia z-scores: {odchylenie_z:.5f} (powinno być ok. 1)")
```

```python
#zadanie 19

prawdopodobienstwa = [0.23, 0.67, 0.89, 0.34, 0.91, 0.12, 0.78, 0.45]
result = [0, 1, 1, 0, 1, 0, 1, 0]

threshold = [0.5, 0.3, 0.7]

wyniki = []

for i in threshold:
  test = []
  for x in prawdopodobienstwa:
    if x >= i:
      outcome = 1
    else:
      outcome = 0
    test.append(outcome)

  liczba_poprawnych = 0
  for n in range(0,(len(result))):
    if result[n] == test[n]:
      liczba_poprawnych = liczba_poprawnych + 1
  accuracy_19 = liczba_poprawnych / len(result)
  wyniki.append(accuracy_19)
  print(f"Accuracy dla threshold: {i} wynosi: {accuracy_19:.2f}")
optymalny_wartosc = max(wyniki)
optymalny_indeks = wyniki.index(optymalny_wartosc)
print(f"Optymalny bedzie threshold {threshold[optymalny_indeks]}")
```
